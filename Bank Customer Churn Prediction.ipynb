```python
from sklearn import svm
from sklearn.metrics import*
from sklearn.feature_selection import chi2,SelectPercentile,SelectKBest,SelectFromModel
from sklearn.impute import SimpleImputer
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier,plot_tree
from sklearn.ensemble import RandomForestClassifier,GradientBoostingClassifier
```

```python
data=pd.read_csv('/content/Churn_Modelling.csv')
data=pd.DataFrame(data)
data
```

```python
null_sum=data.isnull().sum()
null_sum
```

```python
data=data.drop(['RowNumber',	'CustomerId',	'Surname'],axis=1)
data
```

```python
data=pd.get_dummies(data,drop_first=True)
```

```python
data['Exited'].value_counts()
```

```python

X=data.drop('Exited',axis=1)

Y=data['Exited']
```

```python
from imblearn.over_sampling import SMOTE
x_res,y_res=SMOTE().fit_resample(X,Y)
y_res.value_counts()
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(x_res,y_res,test_size=0.2,random_state=42)
```

```python
st=StandardScaler()
X_train=st.fit_transform(X_train)
X_test=st.transform(X_test)
```

```python
LR=LogisticRegression()
LR.fit(X_train,Y_train)
y_pred1=LR.predict(X_test)
print( f"precision_score : { precision_score(Y_test,y_pred1)}")
print(f"accuracy_score   :{accuracy_score(Y_test,y_pred1)}")
print(f"recall_score     :{recall_score(Y_test,y_pred1)}")
print(f"f1_score         :{f1_score(Y_test,y_pred1)}")
print(f"roc_auc_score    :{roc_auc_score(Y_test,y_pred1)}")
print(f"confusion_matrix :\n {confusion_matrix(Y_test,y_pred1)}")
print(f"classification_report :\n {classification_report(Y_test,y_pred1)}")

```

```python
svm=svm.SVC()
svm.fit(X_train,Y_train)
y_pred2=svm.predict(X_test)
print( f"precision_score : { precision_score(Y_test,y_pred2)}")
print(f"accuracy_score   :{accuracy_score(Y_test,y_pred2)}")
print(f"recall_score     :{recall_score(Y_test,y_pred2)}")
print(f"f1_score         :{f1_score(Y_test,y_pred2)}")
print(f"roc_auc_score    :{roc_auc_score(Y_test,y_pred2)}")
print(f"confusion_matrix :\n {confusion_matrix(Y_test,y_pred2)}")
print(f"classification_report :\n {classification_report(Y_test,y_pred2)}")
```

```python
knn=KNeighborsClassifier()
knn.fit(X_train,Y_train)
y_pred3=knn.predict(X_test)
print( f"precision_score : { precision_score(Y_test,y_pred3)}")
print(f"accuracy_score   :{accuracy_score(Y_test,y_pred3)}")
print(f"recall_score     :{recall_score(Y_test,y_pred3)}")
print(f"f1_score         :{f1_score(Y_test,y_pred3)}")
print(f"roc_auc_score    :{roc_auc_score(Y_test,y_pred3)}")
print(f"confusion_matrix :\n {confusion_matrix(Y_test,y_pred3)}")
print(f"classification_report :\n {classification_report(Y_test,y_pred3)}")
```
